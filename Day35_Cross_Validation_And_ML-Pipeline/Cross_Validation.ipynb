# Cross Validation in Machine Learning

```python
import pandas as pd
```

```python
data = pd.read_csv("placement.csv")
data.head(3)
```

```python
data.shape
```

```python
x = data[["cgpa"]]
y = data["package"]
```

```python
df = data.head(5)
```

```python
x_new = df[["cgpa"]]
y_new = df["package"]
```

```python
from sklearn.model_selection import KFold , LeaveOneOut , LeavePOut
```

```python
lo =  LeaveOneOut()
for train , test in lo.split(x_new,y_new):
    print(train,test)
```

```python
lpo =  LeavePOut(p=2)
for train , test in lpo.split(x_new,y_new):
    print(train,test)
```

```python
kf = KFold(n_splits=3)
for train , test in kf.split(x_new,y_new):
    print(train,test)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import cross_val_score

p = cross_val_score(LinearRegression(),x,y,cv=5)
```

```python
p.sort()
```

```python
print(p*100)
```
